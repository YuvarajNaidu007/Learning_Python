Python Syntax

```python
# Indention should be consistent for block of code
if 5 > 2:
    print("Five is greater than two!")
```

```python
# Variables in Python
x = 5 # Interger Variable
y = "Hello, World!" # String Variable

print(x) # Printing the integer variable
print(y) # Printing the string variable
```

Python Statements

```python
# Single statements
print("Python is fun")
```

```python
# Multiple statements
print("Hello World!")
print("Have a good day.")
print("Learning Python is fun!")
```

```python
# Semicolons (Optional, Rarely Used)
# Two statements on the same line without a separator (newline or ;), Python will give an error
print("Hello"); print("How are you?"); print("Bye bye!")
```

Python Outputs - Text

```python
# Python must be inside quotes (" double quotes or ' single quotes)
print("This will work!")
print('This will also work!')
```

```python
# Print (Output) Without a New Line
print("Hello World!", end = " ")
print("I will print on the same line.")
```
